```python
from transaksi.diskon import hitung_diskon
from transaksi.pajak import hitung_ppn

harga_awal = 1000000
diskon_persen = 10

besaran_potongan = hitung_diskon(harga_awal, diskon_persen)
harga_setelah_diskon = harga_awal - besaran_potongan
besaran_ppn = hitung_ppn(harga_setelah_diskon)
total_bayar = harga_setelah_diskon + besaran_ppn

print("1. Harga Awal:", harga_awal)
print("2. Besaran Potongan Diskon:", besaran_potongan)
print("3. Harga Setelah Diskon:", harga_setelah_diskon)
print("4. Besaran PPN:", besaran_ppn)
print("5. Total yang harus dibayar:", total_bayar)
```
